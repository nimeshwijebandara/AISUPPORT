```python
from pathlib import Path
import os


BASE_DIR = Path(r"D:\AI_IT_HELPDESK")

if not BASE_DIR.exists():
    raise FileNotFoundError(
        "Project folder not found: " + str(BASE_DIR)
    )

os.chdir(BASE_DIR)

print("Current Project Directory:", os.getcwd())
print("Project directory verified!")
```

```python
DB_FOLDER = BASE_DIR / "database"

DB_FOLDER.mkdir(parents=True, exist_ok=True)

DB_PATH = DB_FOLDER / "helpdesk.db"

print("Database Folder:", DB_FOLDER)
print("Database File:", DB_PATH)
```

```python
import sqlite3
conn = sqlite3.connect(DB_PATH)
conn.execute("PRAGMA foreign_keys = ON")
cursor = conn.cursor()
cursor.executescript("""

-- 1. Users Table
CREATE TABLE IF NOT EXISTS users (
    user_id INTEGER PRIMARY KEY AUTOINCREMENT,
    username TEXT NOT NULL UNIQUE,
    email TEXT UNIQUE,
    password_hash TEXT,
    role TEXT DEFAULT 'user',
    created_at DATETIME DEFAULT CURRENT_TIMESTAMP
);

-- 2. Categories Table
CREATE TABLE IF NOT EXISTS categories (
    category_id INTEGER PRIMARY KEY AUTOINCREMENT,
    category_name TEXT NOT NULL UNIQUE,
    description TEXT
);

-- 3. Tickets Table
CREATE TABLE IF NOT EXISTS tickets (
    ticket_id INTEGER PRIMARY KEY AUTOINCREMENT,
    user_id INTEGER,
    title TEXT NOT NULL,
    description TEXT,
    category_id INTEGER,
    priority TEXT DEFAULT 'Medium',
    status TEXT DEFAULT 'Open',
    created_at DATETIME DEFAULT CURRENT_TIMESTAMP,
    updated_at DATETIME DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (user_id) REFERENCES users(user_id),
    FOREIGN KEY (category_id)
        REFERENCES categories(category_id)
);

-- 4. Knowledge Base Table
CREATE TABLE IF NOT EXISTS knowledge_base (
    article_id INTEGER PRIMARY KEY AUTOINCREMENT,
    category_id INTEGER,
    problem TEXT NOT NULL,
    solution TEXT NOT NULL,
    source TEXT,
    created_at DATETIME DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (category_id)
        REFERENCES categories(category_id)
);

-- 5. Conversations Table
CREATE TABLE IF NOT EXISTS conversations (
    conversation_id INTEGER PRIMARY KEY AUTOINCREMENT,
    user_id INTEGER,
    ticket_id INTEGER,
    user_message TEXT,
    ai_response TEXT,
    created_at DATETIME DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (user_id) REFERENCES users(user_id),
    FOREIGN KEY (ticket_id) REFERENCES tickets(ticket_id)
);

-- 6. Audit Logs Table
CREATE TABLE IF NOT EXISTS audit_logs (
    log_id INTEGER PRIMARY KEY AUTOINCREMENT,
    user_id INTEGER,
    ticket_id INTEGER,
    action TEXT NOT NULL,
    details TEXT,
    created_at DATETIME DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (user_id) REFERENCES users(user_id),
    FOREIGN KEY (ticket_id) REFERENCES tickets(ticket_id)
);

""")

conn.commit()
conn.close()

print("SUCCESS: SQLite database created!")
print("Location:", DB_PATH)
```

```python
import sqlite3

conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

categories = [
    ("Account & Password", "Login and account problems"),
    ("Network & VPN", "Network and VPN connectivity"),
    ("Printer", "Printer and printing problems"),
    ("Software", "Software installation and errors"),
    ("System Health", "CPU, RAM and disk performance")
]

cursor.executemany("""
INSERT OR IGNORE INTO categories
(category_name, description)
VALUES (?, ?)
""", categories)

conn.commit()
conn.close()

print("IT categories inserted successfully!")
```

```python
conn = sqlite3.connect(DB_PATH)

cursor = conn.cursor()

cursor.execute("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name
""")

tables = cursor.fetchall()

print("DATABASE TABLES")
print("-" * 35)

for table in tables:
    print("[OK]", table[0])

conn.close()
```
